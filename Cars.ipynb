```python
import os
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import xml.etree.ElementTree as ET
```

```python
DATA_PATH = "/content/drive/MyDrive/images"

images = []
labels = []

for img_name in os.listdir(DATA_PATH):
    img_path = os.path.join(DATA_PATH, img_name)
    img = cv2.imread(img_path)
    if img is None:
        continue
    img = cv2.resize(img, (128, 64))
    images.append(img)
    labels.append(1)

images = np.array(images)
labels = np.array(labels)

print("Number of images:", len(images))
```

```python
model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(64,128,3)),
    layers.MaxPooling2D(),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
model.fit(images/255.0, labels, epochs=5, batch_size=8)
```

```python
test_img = cv2.imread("/content/drive/MyDrive/images/Cars0.png")
test_img = cv2.resize(test_img, (128,64))
test_input = np.expand_dims(test_img/255.0, axis=0)

pred = model.predict(test_input)[0][0]

if pred > 0.5:
    print("Plate detected")
else:
    print("No plate")
```

```python
import os
import cv2
import easyocr
import xml.etree.ElementTree as ET
import matplotlib.pyplot as plt

reader = easyocr.Reader(['en'])
image_folder = "/content/drive/MyDrive/images"
annotation_folder = "/content/drive/MyDrive/annotations"

for ann_file in os.listdir(annotation_folder):
    if not ann_file.endswith(".xml"):
        continue  # تجاهل أي مجلد أو ملف مش XML

    xml_path = os.path.join(annotation_folder, ann_file)
    tree = ET.parse(xml_path)
    root = tree.getroot()

    # اسم الصورة مرتبط بالـ XML
    img_name = root.find('filename').text
    img_path = os.path.join(image_folder, img_name)
    img = cv2.imread(img_path)
    if img is None:
        continue

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # استخدام الـ Bounding Box الحقيقي
    for obj in root.findall('object'):
        bbox = obj.find('bndbox')
        x1 = int(bbox.find('xmin').text)
        y1 = int(bbox.find('ymin').text)
        x2 = int(bbox.find('xmax').text)
        y2 = int(bbox.find('ymax').text)

        # قص اللوحة
        plate = img_rgb[y1:y2, x1:x2]

        # قراءة الرقم
        result = reader.readtext(plate, detail=0)
        text = result[0] if len(result) > 0 else "No text"

        # رسم المربع وكتابة الرقم
        cv2.rectangle(img_rgb, (x1, y1), (x2, y2), (255,0,0), 2)
        cv2.putText(img_rgb, text, (x1, y1-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255,0,0), 2)

    # عرض الصورة
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.show()
    print(f"Plate Number ({img_name}):", text)
```

```python
%%writefile app.py

from flask import Flask, request, render_template, send_file
import cv2
import numpy as np
from tensorflow.keras.models import load_model
import easyocr
from PIL import Image
import os
import io
import base64

app = Flask(__name__)
app.config['UPLOAD_FOLDER'] = 'uploads'

# Load the pre-trained car plate detection model
model = load_model('car_plate_detector_model.h5')

# Initialize EasyOCR reader
# This might take some time on the first run
reader = easyocr.Reader(['en'])

@app.route('/')
def index():
    return render_template('index.html')

@app.route('/predict', methods=['POST'])
def predict():
    if 'file' not in request.files:
        return render_template('index.html', message='No file part')
    file = request.files['file']
    if file.filename == '':
        return render_template('index.html', message='No selected file')
    if file:
        # Read the image
        image_bytes = file.read()
        image = Image.open(io.BytesIO(image_bytes))
        img_array = np.array(image)

        # Convert to OpenCV format (BGR)
        img_bgr = cv2.cvtColor(img_array, cv2.COLOR_RGB2BGR)


        # Resize image for model prediction
        resized_img = cv2.resize(img_bgr, (128, 128))
        test_input = np.expand_dims(resized_img / 255.0, axis=0)

        # Predict if a plate is detected
        prediction = model.predict(test_input)[0][0]

        detected_plate_text = "No plate detected."
        img_output_display = img_array.copy()

        if prediction > 0.5:
            # Use the original image for OCR to avoid quality loss from initial resize
            h, w, _ = img_bgr.shape
            x1, y1 = int(w * 0.3), int(h * 0.4)
            x2, y2 = int(w * 0.7), int(h * 0.6)

            plate_roi = img_bgr[y1:y2, x1:x2]

            # Convert plate_roi to RGB for EasyOCR
            plate_roi_rgb = cv2.cvtColor(plate_roi, cv2.COLOR_BGR2RGB)

            # Read text from the plate
            result = reader.readtext(plate_roi_rgb, detail=0)
            plate_text = result[0] if len(result) > 0 else "No text found"
            detected_plate_text = f"Extracted Plate Number: {plate_text}"

            # Draw rectangle and text on the image for display
            cv2.rectangle(img_output_display, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(img_output_display, plate_text, (x1, y1 - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 0, 0), 2)

        # Convert the processed image to base64 for embedding in HTML
        img_output_pil = Image.fromarray(img_output_display)
        buffered = io.BytesIO()
        img_output_pil.save(buffered, format="PNG")
        img_str = base64.b64encode(buffered.getvalue()).decode("utf-8")

        return render_template('index.html',
                               message=detected_plate_text,
                               image_data=img_str)

if __name__ == '__main__':
    # Create uploads directory if it doesn't exist
    os.makedirs(app.config['UPLOAD_FOLDER'], exist_ok=True)
    app.run(host='0.0.0.0', port=5000)
```
